# Games Classes Before Implementing Monte-Carlo Tree Search

```python
from abc import ABC, abstractmethod
```

## Game Parent Class

```python
class Game(ABC):
    num_actions: int

    def __init__(self, history: list[int] | None = None):
        # Initialise a game with history and board representation
        self.board = self._empty_board()
        self.history = []
        if history:
            for action in history:
                self.move(action)
        
    def whose_turn(self) -> int:
        # First player is +1, second player is -1.
        return (1 - len(self.history) % 2) * 2 - 1
    
    def legal_actions(self) -> list[int]:
        # Determine the legal actions.
        if self.terminal():
            return []
        else:
            return self._legal_actions()

    def move(self, action: int):
        # Add a move to the game.
        if action in self.legal_actions():
            player = self.whose_turn()
            self._place_action_on_board(action, player)
            self.history.append(action)
        else:
            raise ValueError(f"Illegal move played! The action {action} is not allowed in position {self.history}!")

    def terminal_result(self, player: int) -> int:
        # Result for given player: +1 win, -1 loss, 0 draw; raises if game isn't over.
        if not self.terminal():
            raise ValueError("The game isn't over, so there can't be a result!")
        else: 
            return player * self._terminal_result()
    
    @abstractmethod
    def _legal_actions(self) -> list[int]:
        # No need to check if game is over here.
        pass

    @abstractmethod
    def _place_action_on_board(self, action: int, player: int):
        pass
    
    @abstractmethod
    def _terminal_result(self) -> int:
        # Result for first player: +1 win, -1 loss, 0 draw
        pass
             
    @abstractmethod
    def _empty_board(self):
        pass

    @abstractmethod
    def terminal(self) -> bool:
        # Termination rules.
        pass
        
```
